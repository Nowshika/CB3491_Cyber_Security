```python
# 🔐 Visualizing Global Cybersecurity Threats (2015–2024)
# Author: Nowshika Mirza R
# Project: Mini Project - Cyber Threat Visualization

# ===========================
# Step 1: Import Libraries
# ===========================
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# ===========================
# Step 2: Load Dataset
# ===========================
df = pd.read_csv('Global_Cybersecurity_Threats_2015-2024.csv')

# Quick look at the data
print("🔍 Dataset Overview:")
print(df.head(), "\n")
print("📊 Columns:", df.columns.tolist())

# ===========================
# Step 3: Clean & Prepare Data
# ===========================
# Convert 'Year' to string for categorical plotting
df['Year'] = df['Year'].astype(str)

# Fill missing values for visualization
df['Attack Type'] = df['Attack Type'].fillna('Unknown')
df['Severity'] = df.get('Severity', pd.Series()).fillna('Unknown')  # If 'Severity' exists
df['Target Country'] = df.get('Target Country', pd.Series()).fillna('Unknown')  # If exists

# ===========================
# Step 4: Financial Loss Over Years by Attack Type (Interactive)
# ===========================
fig1 = px.bar(
    df.groupby(['Year', 'Attack Type'])['Financial Loss (in Million $)'].sum().reset_index(),
    x='Year', y='Financial Loss (in Million $)', color='Attack Type',
    title='Financial Loss Over Years by Attack Type',
    labels={'Financial Loss (in Million $)': 'Loss (Million $)', 'Year': 'Year'},
    barmode='group', height=450
)
fig1.update_layout(xaxis_tickangle=-45, title_x=0.5)
fig1.show()

# ===========================
# Step 5: Number of Affected Users by Target Industry (Top 10)
# ===========================
industry_affected = df.groupby('Target Industry')['Number of Affected Users'].sum().nlargest(10).reset_index()
fig2 = px.bar(
    industry_affected,
    y='Target Industry', x='Number of Affected Users', orientation='h',
    title='Top 10 Target Industries by Affected Users',
    labels={'Number of Affected Users': 'Affected Users', 'Target Industry': 'Industry'},
    height=450
)
fig2.update_layout(yaxis={'categoryorder':'total ascending'}, title_x=0.5)
fig2.show()

# ===========================
# Step 6: Average Incident Resolution Time by Attack Source
# ===========================
source_resolution = df.groupby('Attack Source')['Incident Resolution Time (in Hours)'].mean().reset_index()
fig3 = px.bar(
    source_resolution,
    x='Attack Source', y='Incident Resolution Time (in Hours)',
    title='Average Resolution Time by Attack Source',
    labels={'Incident Resolution Time (in Hours)': 'Resolution Time (Hours)', 'Attack Source': 'Source'},
    height=400
)
fig3.update_layout(title_x=0.5)
fig3.show()

# ===========================
# Step 7: Word Cloud of Attack Types
# ===========================
text = ' '.join(df['Attack Type'].astype(str))

plt.figure(figsize=(12, 8))
wordcloud = WordCloud(width=1200, height=800, background_color='white', colormap='inferno').generate(text)
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Common Cyber Attack Types', fontsize=18, fontweight='bold')
plt.show()

print("✅ Visualization Completed — Your dashboard is ready!")
```

```python
# ===========================
# Additional Graphs for Cyber Threat Analysis
# ===========================

import plotly.graph_objects as go

# ===========================
# Plot 8: Attack Types Distribution (Pie Chart)
# ===========================
attack_counts = df['Attack Type'].value_counts().reset_index()
attack_counts.columns = ['Attack Type', 'Count']

fig4 = px.pie(
    attack_counts, values='Count', names='Attack Type',
    title='Attack Type Distribution',
    color_discrete_sequence=px.colors.qualitative.Bold
)
fig4.update_traces(textposition='inside', textinfo='percent+label')
fig4.update_layout(title_x=0.5)
fig4.show()

# ===========================
# Plot 9: Financial Loss by Target Industry (Bar Chart)
# ===========================
industry_loss = df.groupby('Target Industry')['Financial Loss (in Million $)'].sum().nlargest(15).reset_index()
fig5 = px.bar(
    industry_loss,
    x='Target Industry', y='Financial Loss (in Million $)',
    title='Top 15 Industries by Financial Loss',
    labels={'Financial Loss (in Million $)': 'Loss (Million $)', 'Target Industry': 'Industry'},
    color='Financial Loss (in Million $)',
    color_continuous_scale='Viridis',
    height=500
)
fig5.update_layout(xaxis_tickangle=-45, title_x=0.5)
fig5.show()

# ===========================
# Plot 10: Incident Resolution Time Distribution (Box Plot by Attack Type)
# ===========================
fig6 = px.box(
    df, x='Attack Type', y='Incident Resolution Time (in Hours)',
    title='Incident Resolution Time by Attack Type',
    labels={'Incident Resolution Time (in Hours)': 'Resolution Time (Hours)', 'Attack Type': 'Type'},
    color='Attack Type',
    points='all',
    height=450
)
fig6.update_layout(xaxis_tickangle=-45, title_x=0.5)
fig6.show()

# ===========================
# Plot 11: Security Vulnerability Types (Horizontal Bar Chart)
# ===========================
vulnerability_counts = df['Security Vulnerability Type'].value_counts().nlargest(12).reset_index()
vulnerability_counts.columns = ['Vulnerability', 'Count']

fig7 = px.bar(
    vulnerability_counts,
    y='Vulnerability', x='Count', orientation='h',
    title='Top Security Vulnerability Types',
    labels={'Count': 'Number of Incidents', 'Vulnerability': 'Type'},
    color='Count',
    color_continuous_scale='Plasma',
    height=450
)
fig7.update_layout(yaxis={'categoryorder':'total ascending'}, title_x=0.5)
fig7.show()

# ===========================
# Plot 12: Defense Mechanism Usage (Donut Chart)
# ===========================
defense_counts = df['Defense Mechanism Used'].value_counts().reset_index()
defense_counts.columns = ['Defense Mechanism', 'Count']

fig8 = go.Figure(data=[go.Pie(
    labels=defense_counts['Defense Mechanism'],
    values=defense_counts['Count'],
    hole=0.4,
    marker_colors=px.colors.sequential.Blues
)])
fig8.update_layout(
    title_text='Defense Mechanism Usage',
    annotations=[dict(text='Defense', x=0.5, y=0.5, font_size=20, showarrow=False)]
)
fig8.show()
```

```python
import pandas as pd
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('Global_Cybersecurity_Threats_2015-2024.csv')

# Convert Year to string for categorical plotting
df['Year'] = df['Year'].astype(str)

# ===========================
# Advanced Plot 1: Heatmap of Attack Types vs Target Industries
# ===========================
pivot_data = df.pivot_table(index='Attack Type', columns='Target Industry', values='Financial Loss (in Million $)', aggfunc='sum', fill_value=0)
plt.figure(figsize=(16, 8))
sns.heatmap(pivot_data, cmap='YlOrRd', linewidths=0.5, linecolor='gray')
plt.title('Heatmap: Attack Types vs Target Industries (Financial Loss)', fontsize=16, fontweight='bold')
plt.xlabel('Target Industry')
plt.ylabel('Attack Type')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# ===========================
# Advanced Plot 2: Yearly Trend for Ransomware Financial Loss
# ===========================
ransomware_df = df[df['Attack Type'] == 'Ransomware'].groupby('Year')['Financial Loss (in Million $)'].sum().reset_index()
fig = px.line(ransomware_df, x='Year', y='Financial Loss (in Million $)', title='Yearly Financial Loss from Ransomware', markers=True)
fig.update_layout(title_x=0.5, yaxis_title='Loss (Million $)')
fig.show()

# ===========================
# Advanced Plot 3: Correlation Heatmap of Numeric Features
# ===========================
numeric_cols = ['Financial Loss (in Million $)', 'Number of Affected Users', 'Incident Resolution Time (in Hours)']
corr = df[numeric_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Key Numeric Features', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
